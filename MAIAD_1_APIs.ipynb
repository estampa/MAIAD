# Accessing an API

We will use https://replicate.com services through an API. Since we are not running AI tools in colab we don't need a GPU runtime.

To use an API you have to provide some kind of verification, and it is mainly done using **tokens**. Tokens are strings of characters attached to your account in the service, you have to keep them "secret".

⚡ Then you can check the documentation about accessing the API here https://replicate.com/docs/get-started/python.


⚡ Then search for the service that you want to use here https://replicate.com/explore. For this example we are going to use the [Llama language model](https://replicate.com/meta/meta-llama-3-8b-instruct).

## Steps

First we install the library for accessing the API using the pip command.

Notice the ! at the start of the line to indicate this is a command line tool and not python code.

````
!pip install -q replicate
````

We enter the API token to authenticate the requests. Due to being in colab, the code they show in the website doesn't work, so we use a different method.

🛑
```
export REPLICATE_API_TOKEN=<paste-your-token-here>

replicate.run(...)
```

✅
```
replicate = replicate.Client(api_token='<paste-your-token-here>')

replicate.run(...)
```

````
import replicate

replicate = replicate.Client(api_token='<paste-your-token-here>')
````

To make a request to the API using the library we need to check the model paramters. We can check its API usage in the API tab, selecting python as the programming language ([link](https://replicate.com/meta/meta-llama-3-8b-instruct/api?tab=python)). Also, we can check the list of parameters in the [**Schema** section](https://replicate.com/meta/meta-llama-3-8b-instruct/api/schema).

````
input = {
    "system_prompt": "You are a Machine Learning Tutor AI, dedicated to guiding master students in their journey to become proficient machine learning engineers. Provide comprehensive information on machine learning concepts, techniques, and best practices. Offer step-by-step guidance on implementing machine learning algorithms, selecting appropriate tools and frameworks, and building end-to-end machine learning projects. Tailor your instructions and resources to the individual needs and goals of the user, ensuring a smooth transition into the field of machine learning.",
    "prompt": "Explain how to access machine learning APIs",
}

output = replicate.run(
    "meta/meta-llama-3-8b-instruct",
    input=input
)

print(''.join(output))
````

# Finalizing

When you finish working you have to remember to **stop the runtime**, because there is a time limit and to avoid wasting resources. To stop the runtime click Manage Sessions on the Runtime menu. Once the dialog opens click terminate on the current runtime.

> But when you stop the runtime everything you have not saved is ⚠ **lost** ⚠, so be sure to **download** everything you want to keep before stopping it.
