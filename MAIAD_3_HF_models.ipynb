## Pipelines

This library can run different tasks, that they organize in what they call [pipelines](https://huggingface.co/docs/transformers/pipeline_tutorial), and they are based on tasks or input/output format, like DepthEstimation or TextToAudio ([list](https://huggingface.co/docs/transformers/main_classes/pipelines)).

We can create a pipeline with the task name

```
transcriber = pipeline(task="automatic-speech-recognition")
```
or by selecting the model
```
transcriber = pipeline(model="openai/whisper-large-v2")
```
or both.

````
from transformers import pipeline
import torch

device = torch.device('cuda')

oracle = pipeline("visual-question-answering", model="Salesforce/blip-vqa-base", device=device)
````

Once created we can give it an input and it will process it and give us an output. In this case the input is an image and a question.

First we define the image

````
from IPython.display import Image

image_url = "https://huggingface.co/datasets/Narsil/image_dummy/raw/main/parrots.png"
Image(image_url)
````

Then we call the pipeline with the image and the question

````
output = oracle(question="What are they talking about?", image=image_url)
````

The output format varies with the pipeline.

````
print(output)
````

In this case, it is an array ([]) of dictionaries ({}). So to access it we get the attribute 'answer' of the first element.

````
print(output[0]["answer"])
````

You can also **upload** files to be processed by using the upload icon in the file explorer or the upload option of the folder three dots menu. And then we change the image path in the code.

**Tip**: you can get the file path, by hovering over the file, clicking the three dots on the right and selecting "copy path"

````
image_path = "/content/blackhole.jpg"

output = oracle(question="What's inside?", image=image_path)

print(output[0]["answer"])
````
