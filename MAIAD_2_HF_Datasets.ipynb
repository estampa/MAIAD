In the notebook we will learn how to access Huggingface Datasets from code using their library.

Huggingface 🤗 is a company that offers a set of tools to work with machine learning. Among them:
*   Python libraries to run machine learning models (transformers, diffusers)

For now we are going to interact with the [Datasets](https://huggingface.co/datasets) part.



## Accessing the Dataset

First we install the library for accessing the dataset.

```
!pip install -q datasets
```


We can load datsets from huggingface using their name. In this example we are going to load the [textures-color-1k dataset](https://huggingface.co/datasets/dream-textures/textures-color-1k).

```
from datasets import load_dataset

dataset = load_dataset("dream-textures/textures-color-1k")
```

Each dataset has it's own structure, so we first print it.

```
print(dataset)
```

In this case there is a subset called 'train' with 2 features for each of the 1426 entries.

We will start printing the values of the first 5 entries.

```
examples = 5

for idx in range(examples):
  print(f"Dataset #{idx}")
  print("\t", dataset['train']['image'][idx])
  print("\t", dataset['train']['text'][idx])
```

We can see the description of the texture, but the image is an image Object, so we have to display it.

```
for idx in range(examples):
  print(f"Dataset #{idx}")
  display(dataset['train']['image'][idx])
```
